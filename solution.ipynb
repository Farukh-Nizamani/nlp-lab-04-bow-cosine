```python



import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity


```

### Task 1 -- Bag of Words matrix construciton

```python
corpus = [


    "The product performance was amazing and fast.",
    "The service was fast and the performance was great",
    "Terrible customer service and bad performance "


]



sentence_indexes = ["Sentence 1", "Sentence 2", "Sentence 3"]



vectorizer = CountVectorizer(stop_words='english')



# fit() learns the vocab
# transform() converts the document to 'document-term' matrix. each row represents each document and each column represents unique vocab
x = vectorizer.fit_transform(corpus)



# gets the feature names
vocabulary = vectorizer.get_feature_names_out()
```

```python

df = pd.DataFrame(

    x.toarray(),
    columns=vocabulary,
    index=sentence_indexes
    
)


print("\nDocument-Term Matrix:")
df
```

### Task 2 -- Document Search Engine & Relevnce Ranking

```python
documents = [

    "Machine learning algorithms analyze structured data effectively",
    "Deep learning and neural networks excel at processing unstructured data",
    "Natural language processing helps computers understand human language",
    "Python is widely used for machine learning and data science"

]


query = ["machine learning algorithms for data"]



vectorizer = CountVectorizer()
doc_vector = vectorizer.fit_transform(documents).toarray()
query_vector = vectorizer.transform(query).toarray()

similarity_score = cosine_similarity(query_vector, doc_vector)[0]


ranked_documents = sorted(

    zip(documents, similarity_score),
    key=lambda x: x[1],
    reverse=True

)


for rank, (documents, score) in enumerate(ranked_documents, start=1):
    print(f"{rank}. Score: {score:.4f}")
    print(f"    {documents}\n")
```
